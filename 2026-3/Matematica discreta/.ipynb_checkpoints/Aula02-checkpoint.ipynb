### Proposições Lógicas

A lógica formal pode represnetar as afirmações que fazempos em linguagem cotidiana para apresentar fatos dou transmitir informações. Uma proposição (ou declaração) é uma sentença que é falsa ou verdadeira.
    

**Exemplo**

Considere as sentenças: 

* Dez é menor do que sete. 
* Cheyenne é a capital do estado americano de Wyoming
* Ela é muito talentosa.

**Conectivos e Valores Lógicas**

Representamos letras maiúsculas as pproposições, exemplo, A,B,C...

Conectivos lógicos são operadores que unem proposições simples gerando propisições compostas.

$\land$ - conectivo para "e" (conjunção)

$\vee$  - conectivo para "ou" (disjunção)

$\neg$  - conectivo para negação 

$\to$   - conectivo condicional 

$\leftrightarrow$ - conectivo bicondicional 

**Exemplo**

"Eu comprei maça e abacaxi"

A - comprei maça

B - Comprei Abacaxi
